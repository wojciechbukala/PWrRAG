## Query Translation
Correct and redefine user's question to get better results from LLM

Available techniques:
- **Multi-Query** - generate a few versions of the orginal question, try different perespectives.
- **Decomposition** - decompose complex question into smaller and more specified questions
- **Step-back** - take a stap back in the considered problem to catch wider perespective

## Query construction

Change of the question form using **Self-querying** technique. It splits question to two parts:
1) Text sequence for semantic search
2) Metadata filter (ex. year > 2022)

## Retrieval

Proccess of browsing the knowledge base in order to find relevant informations and provide them to the LLM.
